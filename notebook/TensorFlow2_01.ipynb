```python
import tensorflow as tf
import numpy as np
```

```python
mnist = tf.keras.datasets.mnist

(x_train, y_train), (x_test, y_test) = mnist.load_data()
```

```python
x_train, x_test = x_train / 255, x_test / 255

x_train.shape, y_train.shape, x_test.shape, y_test.shape
```

```python
model = tf.keras.models.Sequential([
    # 28x28の元データを784の縦ベクトルに変換
    tf.keras.layers.Flatten(input_shape=(28, 28)),
    # 784x128の行列に変換
    tf.keras.layers.Dense(128, activation="relu"),
    # レイヤーから出力された特徴量に対してランダムにゼロ化する割合を指定
    tf.keras.layers.Dropout(0.2),
    # 128x10の行列に変換
    tf.keras.layers.Dense(10),
    ])
```

```python
prediction = model(x_train[:1]).numpy()
prediction
```

```python
tf.nn.softmax(prediction).numpy()
```

```python
loss_fn = tf.keras.losses.SparseCategoricalCrossentropy(from_logits=True)
loss_fn(y_train[:1], prediction).numpy()
```

```python
model.compile(
    optimizer='adam',
    loss=loss_fn,
    metrics=['accuracy']
)
```

```python
model.fit(x_train, y_train, epochs=5)
```

```python
model.evaluate(x_test, y_test, verbose=2)
```

```python
probaility_model = tf.keras.Sequential(
    [
        model,
        tf.keras.layers.Softmax(),
    ]
)
```

```python
display_num = 20
predict_result = probaility_model(x_test).numpy()
for i, p in enumerate(predict_result[:display_num]):
    print(np.argmax(p), end=', ')
print()
for i, y in enumerate(y_test[:display_num]):
    print(y, end=', ')
```

```python
count = 0
for i, (y, p) in enumerate(zip(y_test, predict_result)):
    if y != np.argmax(p):
        count += 1
print(f'{count / len(y_test) * 100}%')
```
