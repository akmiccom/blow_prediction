## 概要
- Conditional Generative Adversarial Network (CGAN) を使用した深層学習モデルの学習プロセスを実装
- LS-DYNAのデータを利用して、メッシュ形式のデータから学習を行う
- パリソンデータ（樹脂の押出成形に関する情報）を元に外径や板厚の予測を行うことを目指す

## 主な処理内容：
### データ読み込み・前処理
extract_input_data関数とextract_result_data関数では、LS-DYNAから出力されたデータとパリソンデータから情報を抽出し、モデルに必要な形式に変換します。
抽出後、フィルタリング処理（filter_data）を行い、指定された範囲内のデータのみを使用します。

### モデル構築
- 生成器 (Generator) の関数funcgでは、CNNベースのダウンサンプリングとアップサンプリングを使用して、入力データを処理する
- 生成器は、与えられた条件（ここではパリソンデータ）から新しいデータ（予測された板厚）を生成する
- 識別器 (Discriminator) の関数funcdでは、2D畳み込み層を用いた識別器を定義し、生成されたデータが本物か偽物かを判定する

### 損失関数
- lsg関数とlsd関数はそれぞれ生成器と識別器の損失関数を定義する
- これにより、生成器が生成したデータが識別器を欺けるかどうかが判断される
- 生成器の損失関数は、識別器がその出力を本物だと判断する度合いと、実際のデータとの誤差（L1損失）から計算する

### 学習プロセス
- tr関数は、学習ステップを実行し、勾配を計算して更新を行う
- この部分では、TensorFlowのGradientTapeを使い、モデルの学習を行う

### 結果の可視化：
損失関数の値をプロットするlossplot関数が定義されており、各学習ステップごとの損失の推移を可視化

### メイン処理
- プログラムのエントリーポイントでは、パリソンデータとLS-DYNAのデータを読み込み、学習の準備を行う
- 学習の進捗状況は、指定されたステップ数ごとにログ出力される
- 学習完了後、損失関数のプロットやモデルの保存が行われ、ログファイルに詳細な結果が書き出す
